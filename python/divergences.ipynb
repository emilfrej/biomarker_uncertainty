Divergences per fit for the div_experiment.py models, from each `fits_<model>/diagnostics.csv` (written by `diagnose_fits`).
Run `div_experiment.py` with `DIAGNOSE = True` first.

```python
import pandas as pd

from div_experiment import OUT, FIT
```

```python
#one row per fit and model; failed fits have a status but no divergences
diag = pd.concat({m: pd.read_csv(OUT / f'fits_{m}' / 'diagnostics.csv') for m in FIT}, names=['model']).reset_index(level=0)
divs = diag.pivot(index='patient', columns='model', values='divergences')[FIT]
```

```python
#total divergences per model, over the fits that finished
divs.sum().astype(int).rename('total divergences').to_frame().T
```

```python
#divergences per fit: patients with a divergence in any model, or a failed fit
shown = divs[(divs.fillna(0) > 0).any(axis=1) | divs.isna().any(axis=1)]

pd.set_option('display.max_rows', None)

shown.astype('Int64').astype(object).where(shown.notna(), 'failed')

```
