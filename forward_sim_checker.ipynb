```python
import sys
sys.path.append("AT_Model_Comparison/utils")
from odeModels import LotkaVolterraModel


```

Instantiate a simulator

```python
#def a model
import matplotlib

forward_sim = LotkaVolterraModel()

time_resolution = 10

#set params
forward_sim.SetParams( **{'rS': 0.027,
                        'rR': 0.027,
                        'K': 1,
                        'dD': 1.5,
                        'dS': 0.,
                        'dR': 0.,
                        'S0': 0.74,
                        'R0':0.01,
                        'DMax':1})


schedule = [
    [0,   100, 1],  
    [100, 150, 0],   
    [150, 250, 1],
    ]

forward_sim.Simulate(schedule, dt = 0.1)


#run it using settings from r

df = forward_sim.resultsDf
print(df)
df.plot(x="Time", y=["S", "R", "TumourSize"])


```
